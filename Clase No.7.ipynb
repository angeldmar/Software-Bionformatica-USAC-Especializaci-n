## 7.3. Bar Plots

```python
# Example 1 - Average price by carbody

ax = sns.barplot(x="carbody", y="price", data=raw_data)
ax.bar_label(ax.containers[0])

# Notes: 
# 1 - the lines signify the confidence interval
# 2 - Takes mean by default

raw_data[['carbody', 'price']].groupby('carbody', as_index = False).agg({'price':'mean'})
```

```python
# Example 2 - Total price by carbody

ax = sns.barplot(x="carbody", y="price", hue = 'fueltype', data=raw_data, ci=False)
for container in ax.containers:
    ax.bar_label(container)

# Notes: 
# 1 - the lines signify the confidence interval
# 2 - Takes mean by default

raw_data[['carbody', 'fueltype', 'price']].groupby(['carbody', 'fueltype'], as_index = False).agg({'price':'mean'})
```

```python
raw_data.head()
```

```python
# Example 3 - Total price by carbody for multible categories

categories = ['carbody', 'aspiration','doornumber','drivewheel','enginelocation', 'fuelsystem']


for c in categories:
    
    ax = sns.barplot(x=c, y="price", hue = 'fueltype', data=raw_data, ci=False)
    for container in ax.containers:
        ax.bar_label(container)
    plt.title(c)
    plt.show()
```

```python
# Example 4 - Total price by carbody & fueltype
# make sure you change x with y

ax = sns.barplot(x="price", y="carbody", hue = 'fueltype', data=raw_data, orient = 'h', ci=False)
for container in ax.containers:
    ax.bar_label(container)


raw_data[['carbody', 'fueltype', 'price']].groupby(['carbody', 'fueltype'], as_index = False).agg({'price':'mean'})
```

```python
# Example 5 - Total price by carbody & fueltype - CHANGING COLORS
# make sure you change x with y

colors = ["#0BF6CD", '#FAF242']

ax = sns.barplot(x="price", y="carbody", hue = 'fueltype', data=raw_data, orient = 'h', ci=False, palette=colors)
for container in ax.containers:
    ax.bar_label(container)


# Cool Way to pick colours
# https://htmlcolorcodes.com/color-picker/
```

## 7.4. Histograms

```python
# Example 1 - Investigating the distribution of price

x = raw_data['price'].values

sns.distplot(x, color = '#0BF657');

# As you can see, it's a bit imbalance. Right skewd distribution as the mean is to the right
```

```python
# Example 2 - Investigating the distribution of price, adding the mean

x = raw_data['price'].values

sns.distplot(x, color = 'blue');

# Calculating the mean
mean = raw_data['price'].mean()

#ploting the mean
plt.axvline(mean, 0,1, color = 'red')
```

```python
# Example 3 - Investigating the distribution of all fields, adding the mean

# identifying all numeric columns
numerics = ['int16', 'int32', 'int64', 'float16', 'float32', 'float64']
n_variables = raw_data.select_dtypes(include=numerics).columns


for n in n_variables:
    x = raw_data[n].values
    sns.distplot(x, color = 'blue');

    # Calculating the mean
    mean = raw_data[n].mean()

    #ploting the mean
    plt.axvline(mean, 0,1, color = 'red')
    plt.title(n)
    plt.show()
```

# 7.11. Heat Map



- First we need to create a "Dataset" to display on a Heatmap - we will use a correlation dataset
- .corr() is used to find the pairwise correlation of all columns in the dataframe. Any null values are automatically excluded
- The closer to 1 or -1 the better. As one variable increases, the other variable tends to also increase / decrease
- More Info here: https://statisticsbyjim.com/basics/correlations/

```python
pc = raw_data[['price','wheelbase', 'enginesize','horsepower']].corr(method ='pearson')
pc
```

```python
# Example 1 - Heatmap 

pc = raw_data[['price','wheelbase', 'enginesize','horsepower']].corr(method ='pearson')

cols = ['price','wheelbase', 'enginesize','horsepower']

ax = sns.heatmap(pc, annot=True,
                 yticklabels=cols,
                 xticklabels=cols,
                 annot_kws={'size': 50})
```

```python
# Example 2 - Heatmap for PC - changing cmap


# identifying all numeric columns
numerics = ['int16', 'int32', 'int64', 'float16', 'float32', 'float64']
n_variables = raw_data.select_dtypes(include=numerics).columns

pc = raw_data[n_variables].corr(method ='pearson')
cols = n_variables

ax = sns.heatmap(pc, annot=True,
                 yticklabels=cols,
                 xticklabels=cols,
                 annot_kws={'size': 10},
                 cmap="Greens")

# Examples:
# cmap="YlGnBu"
# cmap="Blues"
# cmap="BuPu"
# cmap="Greens"
```
