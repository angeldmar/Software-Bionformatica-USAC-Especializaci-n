## Inciso No. 2
**¿Qué tipo de variables presenta el archivo? ¿El archivo presenta forma de escalar, vector, matriz o es una combinación?**

**Tipos de variables presentes en el archivo**

El archivo contiene tres tipos de variables. El primero es el cualitativo nominal, que agrupa categorías de texto sin ningún orden ni jerarquía entre ellas; un ejemplo es la variable `GENDER`, que toma los valores "M" o "F" para indicar el sexo biológico del paciente. El segundo tipo es el cualitativo ordinal codificado como entero, correspondiente a variables dicotómicas o binarias de sí o no que se representan numéricamente con los valores 1 y 2; aunque pandas las almacena como `int64`, su naturaleza es categórica, como ocurre con `SMOKING`, `FATIGUE` o `LUNG_CANCER`. El tercer tipo es el cuantitativo continuo, que incluye valores numéricos que representan una magnitud real medible, como la variable `AGE`, que indica la edad del paciente en años.

Las columnas de la 3 a la 15 están codificadas bajo el esquema YES = 2 y NO = 1, por lo que, aunque pandas las interprete como enteros, son variables cualitativas binarias.

---

**Forma de los datos: ¿escalar, vector, matriz o combinación?**

El archivo presenta una combinación de las tres formas. Cada celda individual del DataFrame es un escalar, es decir, un único valor numérico o de texto. A un nivel superior, cada columna constituye un vector unidimensional o Serie de pandas, por ejemplo, la columna `AGE` es un vector de 309 valores enteros y la columna `GENDER` es un vector de 309 cadenas de texto. El DataFrame completo, con sus filas y columnas, forma una estructura bidimensional en la que cada fila representa un paciente y cada columna un atributo.

El archivo no responde a una sola forma, sino a una combinación de las tres, escalares contenidos dentro de vectores, y estos a su vez dentro de una matriz.

## Inciso No. 3
**Ciclo for sobre la variable No. 8 (FATIGUE): conteo de fatigados (Yes = 2) y no fatigados (No = 1).**

```python
fatigue_col = 'FATIGUE '

conteo_yes = 0  # Yes = 2
conteo_no  = 0  # No  = 1

for valor in df[fatigue_col]:
    if valor == 2:
        conteo_yes += 1
    elif valor == 1:
        conteo_no += 1

print(f'Conteo de pacientes FATIGADOS   (Yes = 2): {conteo_yes}')
print(f'Conteo de pacientes NO fatigados (No  = 1): {conteo_no}')
print(f'Total de registros recorridos: {conteo_yes + conteo_no}')
```

## Inciso No. 4
**Graficos con matplotlib (plt) y seaborn (sns).**

### Graficos con matplotlib

```python
# Grafico 1 (plt): Histograma de la variable AGE
plt.figure(figsize=(8, 5))
plt.hist(df['AGE'], bins=20, color='steelblue', edgecolor='black')
plt.title('Distribucion de la Edad de los pacientes')
plt.xlabel('Edad (anos)')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

```python
# Grafico 2 (plt): Grafico de barras - Casos de Cancer de pulmon (YES / NO)
conteos_cancer = df['LUNG_CANCER'].value_counts()

plt.figure(figsize=(6, 5))
plt.bar(conteos_cancer.index, conteos_cancer.values,
        color=['tomato', 'mediumseagreen'], edgecolor='black')
plt.title('Casos de Cancer de Pulmon en el dataset')
plt.xlabel('Diagnostico de Cancer de Pulmon')
plt.ylabel('Numero de pacientes')
for i, val in enumerate(conteos_cancer.values):
    plt.text(i, val + 2, str(val), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

### Graficos con seaborn

```python
# Grafico 1 (sns): Countplot - Distribucion de genero por diagnostico de cancer
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='GENDER', hue='LUNG_CANCER',
              palette='Set2', edgecolor='black')
plt.title('Distribucion de Genero segun Diagnostico de Cancer de Pulmon')
plt.xlabel('Genero')
plt.ylabel('Numero de pacientes')
plt.legend(title='Cancer de Pulmon')
plt.tight_layout()
plt.show()
```

```python
# Grafico 2 (sns): Boxplot de la variable EDAD segun diagnostico de cancer de pulmon
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='LUNG_CANCER', y='AGE',
            palette='pastel', width=0.5)
plt.title('Boxplot de Edad por Diagnostico de Cancer de Pulmon')
plt.xlabel('Cancer de Pulmon')
plt.ylabel('Edad (anos)')
plt.tight_layout()
plt.show()
```

**Interpretación del Boxplot de Edad**

El boxplot permite comparar cómo se distribuye la variable `AGE` entre los pacientes con diagnóstico positivo (**YES**) y negativo (**NO**) de cáncer de pulmón. En cuanto a la mediana, ambos grupos presentan un valor similar, ubicado alrededor de los **60 a 65 años**, lo que indica que la edad central de la población estudiada no difiere drásticamente entre quienes tienen el diagnóstico positivo y quienes no.

Respecto al rango intercuartílico (IQR), representado por la caja, el grupo **YES** muestra una amplitud ligeramente mayor, lo que refleja una mayor variabilidad en las edades de los pacientes diagnosticados con cáncer. El grupo **NO**, en cambio, tiende a concentrarse en un rango más estrecho. En cuanto a los bigotes, los del grupo **YES** se extienden hacia edades más jóvenes, es decir, su límite inferior es más bajo, mientras que el límite superior es comparable en ambos grupos, aproximadamente a los **80 años**. También se observan algunos valores atípicos en ambos grupos, correspondientes a pacientes con edades inusualmente bajas respecto al patrón general de su grupo. Como conclusión, aunque ambos grupos abarcan rangos de edad similares, los pacientes con cáncer de pulmón presentan una distribución etaria ligeramente más dispersa. Esto sugiere que la edad por sí sola no constituye un factor discriminante definitivo entre ambos grupos en este dataset; sin embargo, la tendencia indica que los casos positivos incluyen pacientes en un rango de edad más amplio.
