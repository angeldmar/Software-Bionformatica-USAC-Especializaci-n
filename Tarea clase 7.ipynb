```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rawDataCarros = pd.read_csv(r"/Users/gantz/Documents/Posgrado/Bioinformatica/Software y bases de datos biomedicas/car data all together.csv")

print(rawDataCarros.head())
```

### Ejercicio No. 1

```python
datos1 = rawDataCarros[['wheelbase', 'price']].dropna().sort_values('wheelbase')

plt.figure(figsize=(8,5))
plt.plot(datos1['wheelbase'], datos1['price'], marker='o', linestyle='-', color='blue')
plt.title('Relación de distancia entre ejes  y precio')
plt.xlabel('Distancia entre ejes')
plt.ylabel('Precio')
plt.grid(True)
plt.show()
```

```python
datos2 = rawDataCarros[['curbweight', 'horsepower']].dropna().sort_values('curbweight')

plt.figure(figsize=(8,5))
plt.plot(datos2['curbweight'], datos2['horsepower'], marker='s', linestyle='--', color='red')
plt.title('Relación entre peso en vacio y caballos de fuerza')
plt.xlabel('Peso en vacio')
plt.ylabel('Caballos de fuerza')
plt.grid(True)
plt.show()
```

### Ejercicio No. 2

```python
conteo_carbody = rawDataCarros['carbody'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(conteo_carbody.index, conteo_carbody.values, color='green')
plt.title('Frecuencia por tipo de carrocería')
plt.xlabel('Carroceria')
plt.ylabel('Cantidad')
plt.xticks(rotation=20)
plt.grid(axis='y')
plt.show()
```

```python
conteo_fueltype = rawDataCarros['fueltype'].value_counts()

plt.figure(figsize=(8,5))
plt.barh(conteo_fueltype.index, conteo_fueltype.values, color='orange')
plt.title('Frecuencia por tipo de combustible')
plt.xlabel('Cantidad')
plt.ylabel('Tipo de combustible')
plt.yticks(rotation=20)
plt.grid(axis='x')
plt.show()
```

### Ejercicio No. 3

```python
columnas_numericas = rawDataCarros.select_dtypes(include=['int64', 'float64']).columns

for columna in columnas_numericas:
    plt.figure(figsize=(7,4))
    plt.hist(rawDataCarros[columna].dropna(), bins=15, color='yellow', edgecolor='black')
    plt.title(f'Histograma de {columna}')
    plt.xlabel(columna)
    plt.ylabel('Frecuencia')
    plt.grid(alpha=0.3)
    plt.show()
```

### Ejercicio No. 4

El boxplot de price muestra la mediana del precio de los autos y la dispersión de los datos alrededor de ella. La caja representa el rango intercuartílico, donde se encuentra el 50% de los valores centrales. Los bigotes muestran los valores mínimos y máximos no atípicos, mientras que los puntos fuera de ellos representan valores extremos. En este caso, se observa que la distribución de los precios no es completamente simétrica y presenta algunos valores atípicos altos, lo que sugiere la existencia de autos significativamente más caros que la mayoría.



```python
plt.figure(figsize=(8,5))
sns.boxplot(x=rawDataCarros['price'], color='lightblue')
plt.title('BoxPlot de Price')
plt.xlabel('Price')
plt.show()
```
