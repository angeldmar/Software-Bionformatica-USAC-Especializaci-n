### 4.	Crear una función siguiendo las siguientes instrucciones y realizar un ejemplo que corrobore que la función definida si funciona.

```python
def pattern_count(text,pattern):
    count = 0
    for i in range(len(text) - len(pattern) + 1):
        if text[i:i+len(pattern)] == pattern:
            count += 1
    return count

# Prueba
fragmentoADN = "ATGTGTGTGACCCAGTAGTAGT"
patron = "AGT"

prueba = pattern_count(fragmentoADN, patron)
print(prueba)
```
