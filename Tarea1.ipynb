## Inciso 1: Resumen Markdown

Markdown es un lenguaje de marcado ligero para darle formato al texto usando símbolos especiales.

***

* ## Encabezados

Se pueden utilizar con el Símbolo # La cantidad de # índica el nivel del encabezado. Por ejemplo:

# Encabezado 1
## Encabezado 2
### Encabezado 3
#### Encabezado 4

***

* ## Texto en negrita

Se utiliza doble asterisco o doble guion bajo. Por ejemplo:

**Texto en negrita**
__Texto en negrita__

* ## Texto en cursiva

Se utiliza asterisco o guion bajo.
Por ejemplo:
*Texto en cursiva*
_Texto en cursiva_

* ## Enlaces
Se utiliza corchetes. Por ejemplo:
[Google](https://www.google.com)

* ## Título del Enlace
Se agrega entre comillas el título después de la URL de un enlace.
Por ejemplo:
[Google](https://www.google.com "Buscador")

* ## Imagen

Se usa el símbolo ! seguido de la misma estructura que un enlace.
Por ejemplo:
![Imagen](https://www.google.com/images/branding/googlelogo/2x/googlelogo_color_272x92dp.png)

* ## Título de la Imagen

Se agrega al igual que el titulo del enlace, entre comillas.
Por ejemplo:
![Imagen](https://www.google.com/images/branding/googlelogo/2x/googlelogo_color_272x92dp.png "Logo de Google")

* ## listas ordenadas
Se numeran usando números seguidos de punto.
1. Primer elemento
2. Segundo elemento

* ## listas no ordenadas

Se utilizan asterisco, guion o mas.
* Primer elemento
- Segundo elemento
+ Tercer elemento
