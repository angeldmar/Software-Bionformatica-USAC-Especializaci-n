
## Breast Cancer (METABRIC, Nature 2012 & Nat Commun 2016)




Origen del dataset (https://www.cbioportal.org/study/clinicalData?id=brca_metabric)

La base de datos del Consorcio Internacional de Taxonomía Molecular del Cáncer de Mama (METABRIC) es un proyecto Canadá-Reino Unido que contiene datos de secuenciación específica de 1980 muestras primarias de cáncer de mama. Los datos clínicos y genómicos se descargaron de cBioPortal.

El conjunto de datos fue recopilado por el profesor Carlos Caldas del Cambridge Research Institute y el profesor Sam Aparicio del British Columbia Cancer Center en Canadá y publicado en Nature Communications (Pereira et al., 2016). También apareció en varios artículos, incluidos Nature y otros:
- [Associations between genomic stratification of breast cancer and centrally reviewed tumor pathology in the METABRIC cohort](https://www.nature.com/articles/s41523-018-0056-8)
- [Predicting Outcomes of Hormone and Chemotherapy in the Molecular Taxonomy of Breast Cancer International Consortium (METABRIC) Study by Biochemically-inspired Machine Learning](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC5461908/)

```python
# 1. Dos ejemplos de gráficos de líneas editados (título, labels, marcador, tipo de línea, etc).

df =  data_clinica.copy()


# Promedio de edad por etapa tumoral
linea_etapa = (
    df[['tumor_stage', 'age_at_diagnosis']]
    .dropna()
    .groupby('tumor_stage')['age_at_diagnosis']
    .mean()
    .sort_index()
)

# Promedio de supervivencia por grado histológico
linea_grado = (
    df[['neoplasm_histologic_grade', 'overall_survival_months']]
    .dropna()
    .groupby('neoplasm_histologic_grade')['overall_survival_months']
    .mean()
    .sort_index()
)

fig, axes = plt.subplots(1, 2, figsize=(14,5))

axes[0].plot(
    linea_etapa.index,
    linea_etapa.values,
    marker='o',
    linestyle='--',
    linewidth=2
)

axes[0].set_title("Edad promedio según etapa tumoral")
axes[0].set_xlabel("Etapa tumoral")
axes[0].set_ylabel("Edad al diagnóstico")
axes[0].grid(True)

axes[1].plot(
    linea_grado.index,
    linea_grado.values,
    marker='s',
    linestyle='-.',
    linewidth=2
)

axes[1].set_title("Supervivencia promedio según grado histológico")
axes[1].set_xlabel("Grado histológico")
axes[1].set_ylabel("Meses de supervivencia")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
# 2. Dos ejemplos de gráficos de dispersión (scatter) editados.
plt.figure(figsize=(8,5))

plt.scatter(
    df['tumor_size'],
    df['overall_survival_months'],
    alpha=0.6,
    marker='o'
)

plt.title("Tamaño del tumor vs supervivencia")
plt.xlabel("Tamaño del tumor")
plt.ylabel("Meses de supervivencia")
plt.grid(True)

plt.show()

plt.figure(figsize=(8,5))

plt.scatter(
    df['mutation_count'],
    df['age_at_diagnosis'],
    alpha=0.6,
    marker='x'
)

plt.title("Mutaciones vs edad al diagnóstico")
plt.xlabel("Número de mutaciones")
plt.ylabel("Edad al diagnóstico")
plt.grid(True)

plt.show()
```

```python
# Graficos de barras

conteo_cirugia = df['type_of_breast_surgery'].value_counts()

plt.figure(figsize=(8,5))

plt.bar(conteo_cirugia.index, conteo_cirugia.values)

plt.title("Frecuencia por tipo de cirugía de mama")
plt.xlabel("Tipo de cirugía")
plt.ylabel("Cantidad de pacientes")

plt.xticks(rotation=20)
plt.grid(axis='y')

plt.show()


conteo_er = df['er_status'].value_counts()

plt.figure(figsize=(6,5))

plt.bar(conteo_er.index, conteo_er.values)

plt.title("Distribución del receptor de estrógeno")
plt.xlabel("Estado ER")
plt.ylabel("Cantidad de pacientes")

plt.grid(axis='y')

plt.show()
```

```python
# 4. Dos ejemplos de gráficos de pie editados.

pie_supervivencia = df['overall_survival'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    pie_supervivencia.values,
    labels=['Falleció','Vivió'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Proporción de supervivencia")
plt.show()


pie_menopausia = df['inferred_menopausal_state'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    pie_menopausia.values,
    labels=pie_menopausia.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Estado menopáusico")
plt.show()
```
