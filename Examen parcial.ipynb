### Inciso 1 : Enfermedad genética

#### **Fibrosis quistica**

La fibrosis quística es una enfermedad genética autosómica recesiva que afecta principalmente los pulmones, el páncreas y otros órganos. Es causada por mutaciones en el gen CFTR, el cual codifica una proteína que funciona como canal de cloro en las membranas celulares. Cuando esta proteína no funciona correctamente, el transporte de iones y agua a través de las células epiteliales se altera, provocando la acumulación de secreciones espesas en los pulmones y otros tejidos.

La mutación más común asociada a esta enfermedad es conocida como ΔF508 (Delta F508). Esta mutación consiste en la deleción de tres nucleótidos (CTT) en el gen CFTR. Como consecuencia, se pierde el aminoácido fenilalanina en la posición 508 de la proteína, lo que altera el plegamiento de la proteína CFTR y provoca que sea degradada antes de llegar a la membrana celular. Esta alteración impide el funcionamiento normal del canal de cloro y conduce a las manifestaciones clínicas de la enfermedad.

```python
import numpy as np
import pandas as pd

# secuencia normal
controlSecuenciaWild = "ATCATCTTTGGTGTTTCCTATG"

# Es la mutaci[on de interes, delecion del triplete CTT
mutacionObjetivo = "CTT"

#muestras
muestra1 = "ATCATCTTTGGTGTTTCCTATG" # secuencia normal
muestra2 = "ATCATTGGTGTTTCCTATG" # secuencia con mutacion delta F508
muestra3 = "ATCATTTTGGTGTTTCCTACG" # secuencia mutacion aleatorio
muestra4 = "ATCATTTTGGTGTTTCCTATT" # secuencia mutacion aleatoria
muestra5 =  "ATCATTTTGGCCCCCCCTATG" # secuencia mutacion aleatoria
muestra6 =  "ATCAGGGGGGGGTTTCCTATG" # secuencia mutacion aleatoria
muestra7 = "ATCATTGGTGTTTCCTATG"      # secuencia con mutacion delta F508
muestra8 = "ATCATTGGTGTTTCCTATG"      # secuencia con mutacion delta F508
muestra9 = "ATCATTGGTGTTTCCTATG"    # secuencia con mutacion delta F508

```

### Inciso 3: Definición de función

```python
def detecccionFibrosisQuistica(secuenciaPrueba, referencia, mutacion):

    # Convierte la secuencia a mayúsculas para evitar errores
    # si el usuario escribe nucleótidos en minúsculas
    secuenciaPrueba = secuenciaPrueba.upper()
    referencia = referencia.upper()
    mutacion = mutacion.upper()

    # Compara si la secuencia de prueba es exactamente igual
    # a la secuencia de referencia
    if secuenciaPrueba == referencia:
        return "No presenta la mutacion deltaF508"

    # Verifica si la secuencia de prueba es más corta que la referencia
    # exactamente por la longitud de la mutación (3 nucleótidos)
    # Como en este caso que hay una deleción en delta F508
    if len(secuenciaPrueba) == len(referencia) - len(mutacion):

        # Simula la deleción eliminando una sola vez el triplete CTT
        # de la secuencia de referencia
        referencia_con_delecion = referencia.replace(mutacion, "", 1)

        # Compara la secuencia de prueba con la referencia después
        # de eliminar el triplete CTT
        if secuenciaPrueba == referencia_con_delecion:

            # Si ambas secuencias son iguales, se detecta
            # la mutación ΔF508 asociada a fibrosis quística
            return "Presenta la mutación deltaF508"

    # Si no coincide con la referencia normal ni con la referencia
    # con la deleción simulada, entonces no corresponde
    # a la mutación ΔF508 en este modelo
    return "No coincide con la mutación deltaF508"
```
